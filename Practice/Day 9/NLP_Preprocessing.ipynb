```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

Import Libraries

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer

from textblob import TextBlob
```

5 Different Sentences

```python
sentences = [
    "The movie was absolutely amazing and I loved every scene.",
    "This restaurant serves delicious food at affordable prices.",
    "I am learning Natural Language Processing using Python.",
    "The weather is very pleasant today and everyone is happy.",
    "Artificial Intelligence is changing the future of technology."
]
```

Tokenization

```python
print("TOKENIZATION\n")

for sentence in sentences:
    print("Original:", sentence)
    print("Tokens:", word_tokenize(sentence))
    print()
```

Stopword Removal

```python
stop_words = set(stopwords.words('english'))

print("STOPWORD REMOVAL\n")

for sentence in sentences:
    words = word_tokenize(sentence)

    filtered = [word for word in words if word.lower() not in stop_words]

    print("Original:", sentence)
    print("After Stopword Removal:", filtered)
    print()
```

Stemming

```python
stemmer = PorterStemmer()

print("STEMMING\n")

for sentence in sentences:

    words = word_tokenize(sentence)

    stemmed = [stemmer.stem(word) for word in words]

    print("Original:", sentence)
    print("Stemmed:", stemmed)
    print()
```

Lemmatization

```python
lemmatizer = WordNetLemmatizer()

print("LEMMATIZATION\n")

for sentence in sentences:

    words = word_tokenize(sentence)

    lemma = [lemmatizer.lemmatize(word) for word in words]

    print("Original:", sentence)
    print("Lemmatized:", lemma)
    print()
```

TF-IDF

```python
vectorizer = TfidfVectorizer()

tfidf = vectorizer.fit_transform(sentences)

print("Feature Names\n")

print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix\n")

print(tfidf.toarray())
```

Sentiment Analysis

```python
reviews = [
    "The movie was amazing.",
    "I really enjoyed this film.",
    "Worst movie ever.",
    "The acting was horrible.",
    "It was an average movie."
]

for review in reviews:

    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    print(review)
    print("Polarity:", polarity)
    print("Sentiment:", sentiment)
    print()
```
